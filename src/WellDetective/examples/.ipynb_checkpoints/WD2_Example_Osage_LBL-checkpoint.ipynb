```python
# Directory containing Magnetometry data files
s_folderpath = './LBL/'
# List of files to load.
# * This list includes files that will be meshed together, so it doesn't really make sense to load non-overlapping fights in a single go.
Mag_file_list = ['SRVY0-ACQU5-100Hz.csv', 
            'SRVY0-ACQU6-100Hz.csv',
            'SRVY0-ACQU7-100Hz.csv',
            'SRVY0-ACQU8-100Hz.csv',
            'SRVY0-ACQU9-100Hz.csv',
            'SRVY0-ACQU10-100Hz.csv'
           ]
```

```python
# Load and Process all files from file-list
# Will print (to command window) activity, file statistics, and correction values
# Will produce figures:
#     - flight tracks of each file
#     - Heat map of area of all files
LBL_Data = WellDetective.Load_Process_MultipleMagDataFiles(
    s_folderpath=s_folderpath,
    Mag_File_List=Mag_file_list,
    v_plotdata=True,
    v_runchecks=False,
    skip_errors=True
)
```

            -> MAG_COLUMNS
            
        # Defines range of headings associated with main flight directions.
       	-> primary_range, secondary_range = New_data.find_primary_secondary_headings(tolerance=DEFAULT_HEADING_TOLERANCE) 

        # Normalizes Magnetic data based on primary/secondary header direction.
        # User defined method (default = mean())
        -> New_data.auto_normalize_heading_correction(primary_range, secondary_range, mag_col=magcol[0],method=DEFAULT_HEADING_METHOD)
        
        # Removes data not associated iwth a main flight heading
       	-> New_data.remove_turning_data(primary_range, secondary_range)

```python
# Heading Tolerence and Binning
new_heading_tol = 10
primary_range, secondary_range = New_data.find_primary_secondary_headings(tolerance=new_heading_tol)
New_data.remove_turning_data(primary_range, secondary_range) 
```

```python
# Re-new data level .data_filtered

# # Start work on filtered layer
New_data.start_filtering()

# # Clear rows with 0's and NaN values in Coordinates
New_data.data_filtered.dropna(subset=[latcol[0], loncol[0]], inplace=True)

# Remove turning data
New_data.remove_turning_data(primary_range, secondary_range) 
```

```python
# Calculating flight path length and removing short segments
new_gap_distance = WellDetective.DEFAULT_MAX_GAP_DIST  # Enter new value for gap distane
new_min_seg_len = WellDetective.DEFAULT_MIN_SEG_LEN    # Enger new value for minimum segment length

New_data.segment_and_filter_data(max_gap_distance=new_gap_distance,  
                                 min_segment_length=new_min_seg_len, 
                                 lat_col=latcol[0],  
                                 lon_col=loncol[0])
```

```python
# Normalize Mag to mean
New_data.data_filtered[magcol[0]]=New_data.data_filtered[magcol[0]] - New_data.data_filtered[magcol[0]].mean()

# Equalize heading corrections
New_data.auto_equalize_heading_correction(primary_range, secondary_range, mag_col=magcol[0])

# Project coordinates (UTM zone 12)
zone, hemi, epsg = New_data.get_survey_utm_zone(lat_col=latcol[0], lon_col=loncol[0])
New_data.project_coordinates(utm_zone=zone, lat_col=latcol[0], lon_col=loncol[0])
```
